# AgentX — Smart Action Guardian

AgentX is a safety-focused AI agent built with Strands Agents and Strands Decider. It demonstrates reviewing proposed actions before simulated tool calls proceed.

## Features
- Inspect simulated actions and their risk levels.
- Request confirmation before consequential actions.
- Record `proceed`, `ask_first`, or `deny` safety decisions.
- Demonstrate a safety intervention for tool calls.

## Safety note
This prototype uses simulated, in-memory actions. It does not execute real-world actions. A recorded decision does not mean an action was executed.

## Demo overview

The agent has three simulated tools: `inspect_action`, `request_confirmation`, and `record_safety_decision`. The decision guard reviews tool calls. No real-world action is performed by these demo tools.

```python
demo_actions = {
    'ACT-1001': {'description': 'Generate a report from sample data', 'risk': 'low'},
    'ACT-1002': {'description': 'Delete a production database', 'risk': 'high'},
    'ACT-1003': {'description': 'Send a sample notification', 'risk': 'medium'},
}

for action_id, details in demo_actions.items():
    print(action_id, '—', details)
```

## Example safety outcomes

- `proceed`: the configured safety checks allow the proposed simulated call.
- `ask_first`: additional clarification or confirmation is needed.
- `deny`: the proposed action must not proceed.

These are illustrative outcomes, not a claim that the safety layer is infallible.
